```python
import numpy as np
import time
import timeit
import tracemalloc

N = 1_000_000

py_list = list(range(N))
np_arr = np.arange(N)

def loop_double(values):
    result = []
    for x in values:
        result.append(x * 2)
    return result

def numpy_double(values):
    return values * 2

# Tool 1: perf_counter
for label, func, data in (
    ('loop', loop_double, py_list),
    ('numpy', numpy_double, np_arr)
):
    start = time.perf_counter()
    func(data)
    print(label, round(time.perf_counter() - start, 5), 'seconds')

# Tool 2: timeit
n = 20

t_loop = timeit.timeit(
    'loop_double(py_list)',
    globals=globals(),
    number=n
) / n

t_np = timeit.timeit(
    'numpy_double(np_arr)',
    globals=globals(),
    number=n
) / n

print(
    'loop', round(t_loop, 5),
    'numpy', round(t_np, 5),
    'ratio', round(t_loop / t_np, 2)
)

# Tool 3: tracemalloc
for label, make in (
    ('list', lambda: [x * 2 for x in range(N)]),
    ('array', lambda: np.arange(N) * 2)
):
    tracemalloc.start()

    obj = make()

    current, peak = tracemalloc.get_traced_memory()
    tracemalloc.stop()

    print(
        label,
        'current MB', round(current / 1e6, 1),
        'peak MB', round(peak / 1e6, 1)
    )
```

```python
import numpy as np
import timeit

rows = []

for N in [1_000, 100_000, 1_000_000]:

    py_list = list(range(N))
    np_arr = np.arange(N)

    n = 100 if N < 1_000_000 else 20

    t_loop = timeit.timeit(
        lambda: loop_double(py_list),
        number=n
    ) / n

    t_np = timeit.timeit(
        lambda: numpy_double(np_arr),
        number=n
    ) / n

    ratio = t_loop / t_np

    rows.append((N, t_loop, t_np, ratio))

print(f"{'N':<12}{'Loop (s)':<15}{'NumPy (s)':<15}{'Ratio':<10}")

for N, t_loop, t_np, ratio in rows:
    print(f"{N:<12}{t_loop:<15.6f}{t_np:<15.6f}{ratio:<10.2f}")
```

```python
import numpy as np
import timeit

N = 1_000_000

py_list = list(range(N))
np_arr = np.arange(N)

n = 20

t_loop = min(
    timeit.repeat(
        lambda: loop_double(py_list),
        number=n,
        repeat=5
    )
) / n

t_np = min(
    timeit.repeat(
        lambda: numpy_double(np_arr),
        number=n,
        repeat=5
    )
) / n

print("Best Loop Time:", t_loop)
print("Best NumPy Time:", t_np)
print("Speed Ratio:", t_loop / t_np)
```

```python
import numpy as np

# Function 1: Using Python loops
def standardise_loop(data):

    data = np.asarray(data, dtype=float)

    rows, cols = data.shape

    result = np.zeros((rows, cols))

    means = []
    stds = []

    # Calculate mean and standard deviation
    for j in range(cols):
        column = data[:, j]

        means.append(column.mean())
        stds.append(column.std())

    # Standardise using nested loops
    for i in range(rows):
        for j in range(cols):
            if stds[j] == 0:
                result[i, j] = 0
            else:
                result[i, j] = (
                    data[i, j] - means[j]
                ) / stds[j]

    return result


# Function 2: Using NumPy broadcasting
def standardise_np(data):

    data = np.asarray(data, dtype=float)

    means = data.mean(axis=0)
    stds = data.std(axis=0)

    stds = np.where(stds == 0, 1, stds)

    return (data - means) / stds
```
